3. Dadas las longitudes de los catetos de un triángulo rectángulo, calcula la longitud de la hipotenusa.


4. Dados tres enteros que representan un cierto número de horas, minutos y segundos, calcula su
equivalente sólo en segundos.

5. Dado un número entero *años*, que representa un número de años, introducido por teclado, mostrar
por pantalla el número de días y segundos que hay en *años*. 

6. Dada una cantidad de tiempo expresada en segundos, calcula su equivalente en horas, minutos y
segundos.

céntimos). Dispone, para efectuar las devoluciones, de billetes de 20€, 10 €, 5 € y monedas de 2 € y 1 €,
de manera que siempre devuelve el mínimo número posible de billetes y monedas. Sabiendo que debe
devolver una cierta cantidad de dinero d, averigua cuántos billetes y monedas devuelve de cada clase
(suponiendo que hay suficientes billetes y monedas de todas las clases).
